# 01 Redis 基础

## 八股问答

### 1. Redis 和 MySQL 的区别？为什么 Redis 更快？

| 维度 | Redis | MySQL |
| --- | --- | --- |
| 数据模型 | Key-Value，常用内存数据结构 | 关系模型，表、行、列 |
| 主要存储 | 内存为主，服务重启可选落盘 | 磁盘为主，内存只做 Buffer Pool |
| 查询方式 | 命令操作，无 SQL 解析 | SQL 解析、优化器、执行计划 |
| 事务能力 | 命令原子性、弱事务 | ACID 事务 |
| 一致性 | 高可用场景多为最终一致 | 可由事务和 MVCC 提供强一致 |
| 适用场景 | 缓存、计数器、排行榜、会话 | 业务主数据、复杂关联查询 |

Redis 更快的原因：

- 数据主要在内存中，热路径没有磁盘 IO。
- Redis 6 之前单线程执行命令，6.0 后支持 IO 多线程，但命令执行仍主要单线程，减少了锁和上下文切换。
- 内置 SDS、哈希表、跳表、压缩列表等紧凑数据结构，减少内存和遍历成本。
- 没有 SQL 解析、优化器和复杂表关联，命令协议更简单。
- 支持 Pipeline，减少网络往返。

注意：Redis 不是“一定比 MySQL 快”。它更快是有前提的，数据量受内存限制，且持久化、
网络、大 Key、内存淘汰都会影响性能。

### 2. Redis 的应用场景有哪些？

| 场景 | 典型用法 | 说明 |
| --- | --- | --- |
| 缓存 | `GET/SET` 热点数据 | 减少数据库压力，控制过期时间 |
| 会话存储 | `SET session:token` | 分布式环境共享登录态 |
| 分布式锁 | `SET key token NX PX` | 需要考虑解锁原子性和过期时间 |
| 计数器 | `INCR` | 访问量、限流、点赞 |
| 排行榜 | `ZADD`、`ZRANGE` | 按分数排序 |
| 消息队列 | `LPUSH/BRPOP`、Stream | 轻量异步解耦，可靠性弱于专业 MQ |
| 发布订阅 | `PUBLISH/SUBSCRIBE` | 广播通知 |
| 去重 | `SADD`、`SETBIT` | 集合和位图 |
| 地理信息 | `GEOADD` | 附近的人、距离计算 |
| UV 统计 | HyperLogLog | 允许近似计数时使用 |

选择 Redis 时要考虑：数据可丢失吗、能否接受最终一致、内存成本是否可控、是否需要持久化和高可用。


## 面试话术

“Redis 是内存型 KV 存储，MySQL 是磁盘型关系数据库。Redis 快是因为热数据在内存、
数据结构紧凑、命令简单而且主要由单线程执行，适合缓存、计数、排行榜等低延迟场景。
但 Redis 不能代替 MySQL，它的持久化、事务和关系查询能力较弱，真实系统通常用 Redis
做加速层，用 MySQL 做权威数据源。选择 Redis 时我会先确认能否接受数据丢失窗口、
内存成本和最终一致性。”
